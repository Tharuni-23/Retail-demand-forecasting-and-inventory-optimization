```python
# Identify the daily sales columns
id_cols = [
    "id",
    "item_id",
    "dept_id",
    "cat_id",
    "store_id",
    "state_id"
]

day_cols = [col for col in sales.columns if col.startswith("d_")]

print("Number of daily columns:", len(day_cols))
print("First 5:", day_cols[:5])
print("Last 5:", day_cols[-5:])
```

```python
sales_long = sales.melt(
    id_vars=id_cols,
    value_vars=day_cols,
    var_name="d",
    value_name="sales"
)

print("Shape:", sales_long.shape)
display(sales_long.head())
```

```python
date_map = calendar[["d", "date"]].copy()
date_map["date"] = pd.to_datetime(date_map["date"])
sales_long = sales_long.merge(
    date_map,
    on="d",
    how="left"
)
print(sales_long.shape)
display(sales_long.head())
```

```python
sales_long.info()
```

```python
series_summary = (
    sales_long
    .groupby(["item_id", "store_id"], as_index=False)
    .agg(
        total_sales=("sales", "sum"),
        avg_daily_sales=("sales", "mean"),
        days=("sales", "count")
    )
    .sort_values("total_sales", ascending=False)
)
display(series_summary.head(20))
```

```python
print("Total product-store series:", len(series_summary))
```

```python
print("Total sales:", sales_long["sales"].sum())
```

```python
product_id = "FOODS_3_090"
store_id = "CA_3"
prophet_data = sales_long[
    (sales_long["item_id"] == product_id) &
    (sales_long["store_id"] == store_id)
].copy()

prophet_data = prophet_data[["date", "sales"]].copy()

prophet_data = prophet_data.rename(
    columns={
        "date": "ds",
        "sales": "y"
    }
)

prophet_data = prophet_data.sort_values("ds").reset_index(drop=True)

print("Shape:", prophet_data.shape)
display(prophet_data.head())
display(prophet_data.tail())
```

```python
print(prophet_data.isnull().sum())
print(prophet_data["ds"].is_monotonic_increasing)
print(prophet_data["y"].describe())
```

```python
import importlib.util
print(
    "Prophet installed:",
    importlib.util.find_spec("prophet") is not None
)
```

```python
from prophet import Prophet
```

```python
print(prophet_data.shape)
display(prophet_data.head())
```

```python
forecast_horizon = 30

train = prophet_data.iloc[:-forecast_horizon].copy()
test = prophet_data.iloc[-forecast_horizon:].copy()

print("Training data:", train.shape)
print("Testing data:", test.shape)

print("\nTraining period:")
print(train["ds"].min(), "to", train["ds"].max())

print("\nTesting period:")
print(test["ds"].min(), "to", test["ds"].max())
```

#Training Prophet

```python
model = Prophet(
    yearly_seasonality=True,
    weekly_seasonality=True,
    daily_seasonality=False,
    seasonality_mode="multiplicative"
)

model.fit(train)
```

```python
future = model.make_future_dataframe(
    periods=forecast_horizon,
    freq="D"
)

forecast = model.predict(future)

print("Forecast shape:", forecast.shape)

display(
    forecast[
        ["ds", "yhat", "yhat_lower", "yhat_upper"]
    ].tail(30)
)
```

```python
fig = model.plot(forecast, figsize=(9, 5))
```

```python
fig2 = model.plot_components(forecast)
```

```python
# Extract the forecast for the 30-day test period
predictions = forecast[
    forecast["ds"].isin(test["ds"])
][["ds", "yhat", "yhat_lower", "yhat_upper"]].copy()

# Merge with actual values
results = test.merge(
    predictions,
    on="ds",
    how="left"
)

display(results.head(10))
```

# Calculate Forecast Accuracy

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

actual = results["y"]
predicted = results["yhat"]

mae = mean_absolute_error(actual, predicted)

rmse = np.sqrt(
    mean_squared_error(actual, predicted)
)

# WMAPE is safer than ordinary MAPE when sales contain zeros
wmape = (
    np.sum(np.abs(actual - predicted))
    / np.sum(np.abs(actual))
) * 100

print(f"MAE   : {mae:.2f}")
print(f"RMSE  : {rmse:.2f}")
print(f"WMAPE : {wmape:.2f}%")
```
